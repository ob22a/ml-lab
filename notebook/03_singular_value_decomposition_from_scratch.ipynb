# Singular Value Decomposition (SVD): Deep Mathematical Derivation & Applications

In our previous explorations, we built a strong foundation in **statistics, covariance matrices, PCA, and orthogonal diagonalization**. This notebook shifts our focus exclusively to the **Singular Value Decomposition (SVD)**. 

The overall objective of this notebook is to build a deep, rigorous understanding of the formula:

$$ A = U \Sigma V^T $$

We will not merely compute it; we will explore **why this decomposition exists, what every component means, how it relates geometrically to PCA, and why it enables powerful applications like image compression.**

## 1. Why SVD? The Dimensionality Problem

Recall that standard eigenvalue decomposition, $Av = \lambda v$, naturally describes square linear transformations. It finds the special directions (eigenvectors $v$) that merely scale (by $\lambda$) when transformed by $A$. 

However, a general data matrix or transformation is often rectangular, $A \in \mathbb{R}^{m \times n}$. 

For instance, if $A \in \mathbb{R}^{3 \times 2}$ and $v \in \mathbb{R}^2$, then the output $Av$ lives in $\mathbb{R}^3$. Because the input and output live in entirely different vector spaces, the equation $Av = \lambda v$ cannot generally hold. 

SVD answers a broader, more powerful question: *What are the important orthogonal directions in the input space, where does the matrix send those directions, and how strongly are they scaled?* 

Geometrically, SVD shows that any linear transformation can be decomposed into three distinct steps:
1. **Rotation / Reflection** in the input space.
2. **Scaling** along the coordinate axes.
3. **Rotation / Reflection** into the output space.


## 2. Why $A^TA$ Appears

Since we cannot take the eigendecomposition of a rectangular matrix $A$, we construct a matrix that *is* square and symmetric. 

Given $A \in \mathbb{R}^{m \times n}$, its transpose is $A^T \in \mathbb{R}^{n \times m}$. Their product, $A^TA$, has dimensions $(n \times m) 	imes (m \times n) = n \times n$. This resulting matrix lives purely in the domain (input space) of $A$.

### Symmetry
We can easily prove $A^TA$ is symmetric by taking its transpose:
$$ (A^TA)^T = A^T(A^T)^T = A^TA $$

### Positive Semidefiniteness (PSD)
Because $A^TA$ is symmetric, it has real eigenvalues and orthogonal eigenvectors. But it has another special property: it is **Positive Semidefinite (PSD)**. This means for any vector $x$, the quadratic form is non-negative:
$$ x^T(A^TA)x = (Ax)^T(Ax) = \|Ax\|^2 \ge 0 $$

$A^TA$ becomes **Positive Definite** (strictly $>0$ for $x \neq 0$) exactly when the null space of $A$ contains only the zero vector—meaning $A$ has full column rank. 

*(Note: The analogous logic applies to $AA^T \in \mathbb{R}^{m \times m}$, which acts in the codomain and is positive definite when $A$ has full row rank.)*


## 3. Optimization Derivation of the Right Singular Vectors

Let's approach SVD as a maximization problem. We want to find the direction $v$ (with unit length $\|v\|=1$) that gets stretched the most when multiplied by $A$. 

We maximize the *squared norm* of the output for mathematical convenience:
$$ \max_{\|v\|=1} \|Av\|^2 $$

Using the identity we found during the PSD proof, we can rewrite this as a constrained optimization problem:
$$ \max_{\|v\|=1} v^TA^TAv $$

To solve this, we introduce the **Lagrangian**, using $\lambda$ as the Lagrange multiplier for the constraint $v^Tv = 1$:
$$ \mathcal{L}(v, \lambda) = v^TA^TAv - \lambda(v^Tv - 1) $$

Taking the gradient with respect to $v$ and setting it to zero (noting that $A^TA$ is symmetric):
$$ \nabla_v (v^TA^TAv) = 2A^TAv $$
$$ 2A^TAv - 2\lambda v = 0 $$
$$ A^TAv = \lambda v $$

This is exactly the eigenvalue equation for $A^TA$! The maximizing direction is the eigenvector associated with the largest eigenvalue. Subsequent constrained maximizations (enforcing orthogonality to the first vector) yield the remaining eigenvectors.


## 4. What the Eigenvalue Means Here

Let's look closer at that eigenvalue $\lambda_i$. For a normalized eigenvector $v_i$ (where $v_i^Tv_i = 1$):
$$ A^TAv_i = \lambda_i v_i $$

If we multiply on the left by $v_i^T$:
$$ v_i^TA^TAv_i = \lambda_i v_i^Tv_i $$
$$ \|Av_i\|^2 = \lambda_i $$

This is a critical distinction: **The eigenvalue $\lambda_i$ is the *squared stretch*, not the actual stretch.**

To find the actual stretch applied by $A$, we take the square root. We define this as the **Singular Value**, $\sigma_i$:
$$ \sigma_i = \sqrt{\lambda_i} = \|Av_i\| $$


## 5. Right and Left Singular Vectors

The orthonormal eigenvectors of $A^TA$ are called the **Right Singular Vectors**, denoted as $v_i$. They live in the input/domain space and identify the special directions along which $A$ acts purely as a scaling operation followed by a change of direction.

We know that $A$ stretches $v_i$ by an amount $\sigma_i$. But what is the *direction* of the output vector? Let's call this normalized output direction $u_i$, the **Left Singular Vector**.

For $\sigma_i > 0$, we define $u_i$ exactly as the normalized output:
$$ u_i = \frac{Av_i}{\sigma_i} $$

By multiplying both sides by $\sigma_i$, we get the foundational relationship of SVD:
$$ Av_i = \sigma_i u_i $$

Unlike eigenvectors, $Av_i$ is generally not parallel to $v_i$; it points in an entirely new direction $u_i$ in the output space.

We can also show the symmetric relationship by multiplying our definition by $A^T$:
$$ A^Tu_i = A^T \left( \frac{Av_i}{\sigma_i} \right) = \frac{A^TAv_i}{\sigma_i} $$
Substitute $A^TAv_i = \lambda_i v_i = \sigma_i^2 v_i$:
$$ A^Tu_i = \frac{\sigma_i^2 v_i}{\sigma_i} = \sigma_i v_i $$


## 6. Proving Orthonormality

We know the right singular vectors $v_i$ are orthonormal because $A^TA$ is a real symmetric matrix. But are the newly constructed left singular vectors $u_i$ also orthonormal?

Let's check the dot product of $u_i$ and $u_j$ for $i \neq j$:
$$ u_i^Tu_j = \left( \frac{Av_i}{\sigma_i} \right)^T \left( \frac{Av_j}{\sigma_j} \right) = \frac{v_i^TA^TAv_j}{\sigma_i\sigma_j} $$

Since $v_j$ is an eigenvector of $A^TA$, we substitute $A^TAv_j = \sigma_j^2 v_j$:
$$ u_i^Tu_j = \frac{v_i^T(\sigma_j^2 v_j)}{\sigma_i\sigma_j} = \frac{\sigma_j^2}{\sigma_i\sigma_j} (v_i^Tv_j) $$

Because $v_i$ and $v_j$ are orthogonal, $v_i^Tv_j = 0$, meaning $u_i^Tu_j = 0$. The left singular vectors are indeed orthogonal! (And for $i=j$, the norm equals 1).


## 7. Deriving the Matrix Form $A = U\Sigma V^T$

We have established the individual relationships for each direction:
$$ Av_i = \sigma_i u_i $$

We can stack all these column vectors into matrices. 
Let $V = [v_1, v_2, \dots, v_n]$ and $U = [u_1, u_2, \dots, u_m]$. 
Let $\Sigma$ be a diagonal matrix of the singular values $\sigma_i$.

We can write all the individual equations simultaneously as a matrix multiplication:
$$ AV = U\Sigma $$

Because $V$ is an orthogonal matrix (its columns are orthonormal bases), its inverse is its transpose ($V^{-1} = V^T$). We multiply both sides on the right by $V^T$:
$$ A = U\Sigma V^T $$

This is the Singular Value Decomposition.


## 8. Geometry, Rank, and the Outer-Product Form

### Geometric Interpretation
Think of applying $A$ to a vector $x$ as three sequential operations: $A = U\Sigma V^T$.
1. **$V^T$**: Rotates the input vector $x$ into the special orthonormal coordinate system (the right singular vectors).
2. **$\Sigma$**: Scales each of these new coordinates independently by $\sigma_i$. (A unit circle becomes an ellipse).
3. **$U$**: Rotates the scaled ellipse into the final output-space orientation.

### Rank and Null Space
The **rank of $A$** is exactly equal to the number of nonzero singular values. 
- Right singular vectors corresponding to $\sigma=0$ get crushed to zero; they span the **null space** of $A$.
- Left singular vectors corresponding to nonzero singular values span the **column space** of $A$.

### The Rank-1 Outer-Product Decomposition
If we expand the matrix multiplication $U\Sigma V^T$, we can rewrite $A$ as a sum of rank-1 matrices (outer products):
$$ A = \sum_{i=1}^{r} \sigma_i u_i v_i^T $$

An outer product $u_i v_i^T$ creates an $m \times n$ matrix of rank 1. 
Therefore, **SVD represents a matrix as a sum of independent rank-1 matrices, ordered from most significant (largest $\sigma$) to least significant.**


## 9. Truncated SVD, Image Compression, and the Eckart-Young-Mirsky Theorem

If we only keep the first $k$ singular values (where $k < r$), we get a **Truncated SVD**:
$$ A_k = \sum_{i=1}^{k} \sigma_i u_i v_i^T = U_k\Sigma_k V_k^T $$

The **Eckart-Young-Mirsky Theorem** states that among all matrices of rank at most $k$, the truncated SVD $A_k$ gives the absolute best approximation of $A$ in the Frobenius norm:
$$ A_k = \arg\min_{\operatorname{rank}(B) \le k} \|A - B\|_F $$

The reconstruction error (MSE) is exactly defined by the singular values we discarded: $\sum_{i=k+1}^r \sigma_i^2$.

### Image Compression
If we treat a grayscale image as a matrix $A \in \mathbb{R}^{m \times n}$, storing it takes $mn$ values.
By using $A_k$, we only need to store $U_k$ ($mk$ values), $\Sigma_k$ ($k$ values), and $V_k$ ($nk$ values).
**Compressed Storage = $k(m + n + 1)$**

## 10. PCA vs SVD Equivalence

In our previous notebooks, we computed PCA using the covariance matrix eigendecomposition:
$$ C = \frac{1}{n-1} X_c^TX_c $$
$$ C = V\Lambda V^T $$

Let's look at SVD on the centered data matrix $X_c = U\Sigma V^T$:
$$ X_c^TX_c = (U\Sigma V^T)^T (U\Sigma V^T) = V\Sigma^T U^T U \Sigma V^T $$
Since $U^TU = I$, this simplifies to:
$$ X_c^TX_c = V\Sigma^2 V^T $$

Substitute this back into the covariance formula:
$$ C = \frac{1}{n-1} (V\Sigma^2 V^T) = V \left( \frac{\Sigma^2}{n-1} \right) V^T $$

Therefore, the principal directions (eigenvectors $V$) are exactly the right singular vectors of $X_c$, and the eigenvalues are directly related to the singular values:
$$ \lambda_i = \frac{\sigma_i^2}{n-1} $$

Computing SVD directly on $X_c$ is often more numerically stable and computationally efficient for high-dimensional data than explicitly forming $X_c^TX_c$. Let's prove their equivalence programmatically on the Digits dataset.

### Computational Complexity & Numerical Stability

You might wonder why we prefer SVD if explicitly calculating the covariance matrix $C = X_c^TX_c$ is sometimes faster for 'tall and skinny' datasets.

The limitations of Covariance-based PCA become extremely apparent with **high-dimensional data** (like images or genomics where $d \gg n$).

Imagine a dataset like Olivetti Faces ($n=400$ samples, $d=4096$ features):
1. **Covariance PCA:** We must form $X_c^TX_c$, a massive $4096 \times 4096$ matrix, requiring millions of floating-point operations just to construct, followed by an immensely slow $O(d^3)$ eigendecomposition. 
2. **Direct SVD:** A 'thin' SVD operates directly on the $400 \times 4096$ matrix. It completely avoids computing the massive covariance matrix $A^TA$ by utilizing optimized numerical approaches (like bidiagonalization and Householder reflections) to solve for $V$, $\Sigma$, and $U$ directly. The computational complexity scales relative to the smaller dimension, making it vastly faster and more memory-efficient.

Furthermore, calculating $X_c^TX_c$ involves multiplying the values in the matrix, which fundamentally squares the condition number. If the data has very tiny variances in some directions, squaring them causes severe **floating-point precision loss** as they round to zero. Direct SVD sidesteps this step entirely, operating directly on $X_c$ and preserving double the numerical precision! This is why professional libraries like `scikit-learn` use SVD under the hood for PCA.
