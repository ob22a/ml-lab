## Section 1: Vector Notation and Column Convention

In machine learning and linear algebra, vectors are conventionally represented as **column vectors** rather than row vectors.

$$ x = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix} $$

Why this convention? Because a matrix $A \in \mathbb{R}^{m \times n}$ maps $x \in \mathbb{R}^n$ to $Ax \in \mathbb{R}^m$. The dimensions align naturally: $(m \times n)(n \times 1) = (m \times 1)$.

The transpose $x^T$ converts a column vector into a row vector. This is why we constantly see transposes in ML formulas like $x^T y$, $A^T A$, etc.

## Section 2: Matrices and Notation

A matrix $A \in \mathbb{R}^{m \times n}$ transforms a vector $x \in \mathbb{R}^n$ into $Ax \in \mathbb{R}^m$.

**Diagonal matrices** are especially important. $D = \text{diag}(d_1, d_2, \dots, d_n)$ represents independent scaling along coordinate axes. 

$$ \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix} \to \begin{bmatrix} d_1 x_1 \\ d_2 x_2 \\ \vdots \\ d_n x_n \end{bmatrix} $$

Later, diagonalization will show that complicated transformations can be converted into this simple scaling form under the correct basis.

```python
D = Matrix([
    [2, 0],
    [0, 3]
])
x = Vector([1, 1])
print("D * x =", D * x) # Scaling x-axis by 2, y-axis by 3
```

## Section 3: Vector Products

### Inner Product / Dot Product
For $x, y \in \mathbb{R}^n$, the inner product is $x^T y = \sum x_i y_i$.
The result is a scalar. Geometrically: $x^T y = \|x\|_2 \|y\|_2 \cos(\theta)$.
If $x^T y = 0$, the vectors are orthogonal (perpendicular).

### Outer Product
The outer product $xy^T$ produces an $m \times n$ matrix (if $x \in \mathbb{R}^m, y \in \mathbb{R}^n$).
This is a rank-1 matrix, crucial for covariance matrices and low-rank approximations.

```python
v1 = Vector([1, 0])
v2 = Vector([0, 1])

dot_product = v1.dot(v2)
print("Dot product of v1 and v2:", dot_product)
print("Are they orthogonal?", v1.is_orthogonal(v2))
```

## Section 4: Norms

A norm $\| \cdot \| : \mathbb{R}^n \to \mathbb{R}$ measures the "size" or "length" of a vector. It satisfies:
1. Non-negativity: $\|x\| \geq 0$
2. Definiteness: $\|x\| = 0 \iff x = 0$
3. Homogeneity: $\|\alpha x\| = |\alpha| \|x\|$
4. Triangle inequality: $\|x + y\| \leq \|x\| + \|y\|$

- **$L_1$ norm**: $\sum |x_i|$ (Manhattan distance, leads to sparsity in ML)
- **$L_2$ norm**: $\sqrt{\sum x_i^2}$ (Euclidean distance, standard error)
- **$L_\infty$ norm**: $\max |x_i|$ (Max absolute error)

```python
v = Vector([3, -4])
print("L1 Norm:", v.norm(1))
print("L2 Norm:", v.norm(2))
print("L-infinity Norm:", v.norm('inf'))

# Distance between two points
u = Vector([0, 0])
dist = (v - u).norm(2)
print("Distance between v and u:", dist)
```

## Section 5: Matrix-Vector Multiplication (Multiple Interpretations)

Given $A \in \mathbb{R}^{m \times n}$ and $x \in \mathbb{R}^n$, we have $y = Ax \in \mathbb{R}^m$.

### Row Interpretation
Each output coordinate is the dot product between a row of $A$ and $x$.
$$ Ax = \begin{bmatrix} a_1^T x \\ a_2^T x \\ \vdots \\ a_m^T x \end{bmatrix} $$

### Column Interpretation
$Ax$ is a linear combination of the columns of $A$.
If $A = [a_1, a_2, \dots, a_n]$, then:
$$ Ax = x_1 a_1 + x_2 a_2 + \dots + x_n a_n $$

Therefore, $Ax$ belongs to the span of the columns of $A$ (the Column Space/Range of A).

```python
A = Matrix([
    [1, 2],
    [3, 4]
])
x = Vector([2, 1])

print("Ax =", A * x)
# This is equivalent to 2*[1, 3] + 1*[2, 4] = [2+2, 6+4] = [4, 10]
```

## Section 8: Linear Independence, Basis, and Dimension

A set of vectors is **linearly independent** if $c_1 v_1 + \dots + c_n v_n = 0$ only has the trivial solution $c_1 = \dots = c_n = 0$.
If they are dependent, one vector is redundant and does not expand the span.

A **basis** is a set of vectors that:
1. Spans the space
2. Is linearly independent

If $\det(A) \neq 0$, the columns form a basis for $\mathbb{R}^n$. If $\det(A) = 0$, at least one dimension is redundant/collapsed.

## Assignment 1: The Restricted Robot

The robot has two allowed movement vectors:
- Move A = (2, 1)
- Move B = (1, -2)

The robot can only use **integer** multiples of these moves. 
We want to reach a target coordinate $(x, y)$. This forms a linear combination equation:
$$ a \begin{bmatrix} 2 \\ 1 \end{bmatrix} + b \begin{bmatrix} 1 \\ -2 \end{bmatrix} = \begin{bmatrix} x \\ y \end{bmatrix} $$

This is exactly a matrix-vector equation $M \vec{c} = \vec{t}$:
$$ \begin{bmatrix} 2 & 1 \\ 1 & -2 \end{bmatrix} \begin{bmatrix} a \\ b \end{bmatrix} = \begin{bmatrix} x \\ y \end{bmatrix} $$

To solve for the coefficients $a$ and $b$, we find the inverse of the matrix $M$.
The determinant is $\det(M) = (2)(-2) - (1)(1) = -4 - 1 = -5$.
The inverse is:
$$ M^{-1} = \frac{1}{-5} \begin{bmatrix} -2 & -1 \\ -1 & 2 \end{bmatrix} = \frac{1}{5} \begin{bmatrix} 2 & 1 \\ 1 & -2 \end{bmatrix} $$

Multiplying both sides by $M^{-1}$:
$$ \begin{bmatrix} a \\ b \end{bmatrix} = \frac{1}{5} \begin{bmatrix} 2 & 1 \\ 1 & -2 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} \frac{2x + y}{5} \\ \frac{x - 2y}{5} \end{bmatrix} $$

Because the robot is restricted to integers, a target $(x, y)$ is only reachable if $(2x + y)$ is a multiple of 5, AND $(x - 2y)$ is a multiple of 5.

## Section 11: Transpose

The transpose $A^T$ swaps the rows and columns of a matrix. Mathematically, $(A^T)_{ij} = A_{ji}$.

A critical property for machine learning: $(AB)^T = B^T A^T$.
Why reverse the order? Because if $B$ acts on a vector first, and then $A$, when we take the transpose we must apply the "un-doing" or transposed operations in the reverse sequence to maintain dimensional consistency.

Expressions like $A^T A$ and $AA^T$ are always symmetric, and are the foundation of least squares and covariance.

```python
A = Matrix([
    [1, 2, 3],
    [4, 5, 6]
])
print("A:\n", A)
print("A^T:\n", A.transpose())

# Let's verify (A^T A)^T = A^T A
At_A = A.transpose() * A
print("(A^T A) is symmetric:", At_A.is_symmetric)
```

## Section 12: Orthogonality & Orthonormality

Vectors $x$ and $y$ are orthogonal if $x^T y = 0$.

A set of vectors $q_1, \dots, q_n$ is **orthonormal** if they are mutually orthogonal and all have a length ($L_2$ norm) of 1. 
Mathematically, we write this using the Kronecker delta:
$$ q_i^T q_j = \delta_{ij} = \begin{cases} 1 & \text{if } i = j \\ 0 & \text{if } i \neq j \end{cases} $$

### Proof: Why the Transpose is the Inverse
Let's pack these orthonormal vectors into a matrix $Q = [q_1, q_2, \dots, q_n]$.
Now, let's explicitly compute the matrix multiplication $Q^T Q$:

$$ Q^T = \begin{bmatrix} q_1^T \\ q_2^T \\ \vdots \\ q_n^T \end{bmatrix} $$

$$ Q^T Q = \begin{bmatrix} q_1^T \\ q_2^T \\ \vdots \\ q_n^T \end{bmatrix} \begin{bmatrix} q_1 & q_2 & \dots & q_n \end{bmatrix} $$

By the row-column rule of matrix multiplication, the entry in the $i$-th row and $j$-th column of the result is exactly the dot product $q_i^T q_j$:

$$ (Q^T Q)_{ij} = q_i^T q_j $$

Because our vectors are orthonormal:
- The diagonal entries ($i=j$) are $q_i^T q_i = 1$.
- The off-diagonal entries ($i \neq j$) are $q_i^T q_j = 0$.

Therefore:
$$ Q^T Q = \begin{bmatrix} 1 & 0 & \dots & 0 \\ 0 & 1 & \dots & 0 \\ \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & \dots & 1 \end{bmatrix} = I $$

If $Q$ is square, then $Q^T Q = I$ implies that **$Q^{-1} = Q^T$**. 
This is why orthogonal matrices are so powerful: inverting them is computationally free!

**Orthogonal matrices preserve lengths and angles!**
$$ \|Qx\|^2 = (Qx)^T (Qx) = x^T Q^T Q x = x^T I x = x^T x = \|x\|^2 $$

## Section 13: Projection

The projection of $x$ onto a vector $u$ is the closest vector in the span of $u$.
$$ \text{proj}_u(x) = \frac{x \cdot u}{u \cdot u} u $$

The residual (error) $r = x - p$ is always orthogonal to the projection.

```python
class ProjectionVisualizer:
    @staticmethod
    def plot_1d_projection(x, u, title="Projection"):
        fig, ax = plt.subplots(figsize=(8, 8))
        
        p = x.project_onto(u)
        r = x - p
        
        # Plot original vectors
        ax.quiver(0, 0, x[0], x[1], angles='xy', scale_units='xy', scale=1, color='blue', label='x (Original)')
        ax.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color='red', alpha=0.5, label='u (Target)')
        
        # Plot projected vector
        ax.quiver(0, 0, p[0], p[1], angles='xy', scale_units='xy', scale=1, color='green', linewidth=2, label='p (Projection)')
        
        # Plot residual vector using a standard dashed line (avoids matplotlib quiver bugs)
        ax.plot([p[0], x[0]], [p[1], x[1]], color='orange', linestyle='--', linewidth=2, label='r = x - p (Orthogonal)')
        
        # Plot span of u
        max_val = max(abs(x[0]), abs(x[1]), abs(u[0]), abs(u[1])) * 1.5
        span_factor = max_val / (u.norm() + 1e-9)
        u_ext_pos, u_ext_neg = u * span_factor, u * -span_factor
        ax.plot([u_ext_neg[0], u_ext_pos[0]], [u_ext_neg[1], u_ext_pos[1]], color='gray', linestyle=':', alpha=0.5, label='Span(u)')
        
        ax.set_xlim(-max_val, max_val)
        ax.set_ylim(-max_val, max_val)
        ax.axhline(0, color='black')
        ax.axvline(0, color='black')
        ax.set_aspect('equal')
        ax.set_title(title)
        ax.legend()
        plt.show()

# Let's use a highly educational example:
x = Vector([4, 3])
u = Vector([5, 0])

ProjectionVisualizer.plot_1d_projection(x, u, title="Projection: Extracting the X-component")
```

## Section 15: Rank

The **rank** of a matrix is the number of linearly independent rows (or columns). It represents the number of independent directions preserved or generated by the matrix.

- A full-rank $2 \times 2$ matrix transforms a plane to a plane.
- A rank 1 matrix collapses the plane into a line.
- A rank 0 matrix collapses everything to a point.

```python

# Full rank matrix
A_full = Matrix([
    [1, 0],
    [0, 1]
])
print("Rank of A_full:", A_full.rank())

# Rank deficient matrix (second row is 2x the first)
A_deficient = Matrix([
    [1, 2],
    [2, 4]
])
print("Rank of A_deficient:", A_deficient.rank())
```

## Section 16: Determinants (Deep Geometric Interpretation & Properties)

The determinant $\det(A)$ of a square matrix represents the factor by which an $n$-dimensional volume is scaled under the transformation $A$.

- $\det(A) = 2$: Area/volume is doubled.
- $\det(A) = -1$: Area is preserved, but orientation is reversed (e.g., a reflection).
- $\det(A) = 0$: The volume is compressed to 0. A dimension was lost! (This means rank deficiency and linear dependence).

### Computational Properties of Determinants
Determinants possess incredibly beautiful algebraic properties that simplify complex matrix math. For any $n \times n$ matrices $A$ and $B$:

1. **Multiplicative Property:** $\det(AB) = \det(A)\det(B)$. (The scaling factor of two transformations composed is the product of their individual scaling factors).
2. **Transpose Property:** $\det(A^T) = \det(A)$. (Transposing a matrix doesn't change how much it scales space).
3. **Inverse Property:** $\det(A^{-1}) = \frac{1}{\det(A)}$. (If a matrix scales space by 2, its inverse must scale space by $1/2$ to undo it).
4. **Scalar Multiplication:** $\det(cA) = c^n \det(A)$. (Because scaling every column by $c$ scales an $n$-dimensional volume by $c^n$).
5. **Diagonal/Triangular Matrices:** The determinant is simply the product of the diagonal entries: $\det(D) = d_{11} \times d_{22} \times \dots \times d_{nn}$.
6. **Row/Column Swaps:** Swapping any two rows (or any two columns) of a matrix multiplies the determinant by $-1$. This represents flipping the orientation of the space!
7. **Row Addition:** Adding a multiple of one row to another row (or column to column) **does not change the determinant**. This is the fundamental reason why we can use Gaussian elimination to calculate determinants!

### Proof: The Determinant is the Product of Eigenvalues
One of the most important theorems in Linear Algebra is that $\det(A) = \prod_{i=1}^n \lambda_i$.

*(Note: This proof relies on the diagonalization property $A = V \Lambda V^{-1}$, which we will formally derive in Section 23. We are using it here to show how powerful the multiplicative property of determinants is!)*

**Proof:**
Start with the diagonalization equation:
$$ A = V \Lambda V^{-1} $$

Take the determinant of both sides:
$$ \det(A) = \det(V \Lambda V^{-1}) $$

Apply the multiplicative property (Rule 1):
$$ \det(A) = \det(V) \det(\Lambda) \det(V^{-1}) $$

Because multiplication of scalars is commutative, we can reorder the terms on the right side:
$$ \det(A) = \det(\Lambda) \det(V) \det(V^{-1}) $$

Apply the multiplicative property in reverse to the $V$ terms:
$$ \det(V) \det(V^{-1}) = \det(V V^{-1}) $$

Since $V V^{-1} = I$, and the determinant of the identity matrix is $1$:
$$ \det(A) = \det(\Lambda) \times 1 $$

Finally, because $\Lambda$ is a diagonal matrix containing the eigenvalues $\lambda_i$ on its diagonal, we use Rule 5:
$$ \det(A) = \lambda_1 \times \lambda_2 \times \dots \times \lambda_n = \prod_{i=1}^n \lambda_i $$
**Q.E.D.**

Let's visualize how matrices transform a unit square to see this area scaling in action.

```python
class TransformationVisualizer:
    @staticmethod
    def plot_2d_transformation(matrix, title="Transformation & Determinant"):
        import matplotlib.patches as patches
        fig, ax = plt.subplots(figsize=(8, 8))
        
        # Standard basis e1, e2
        e1, e2 = Vector([1, 0]), Vector([0, 1])
        Ae1, Ae2 = matrix * e1, matrix * e2
        
        # The original unit square
        square = patches.Polygon([[0,0], [1,0], [1,1], [0,1]], closed=True, alpha=0.2, color='blue', label='Unit Square (Area=1)')
        ax.add_patch(square)
        
        # The transformed parallelogram
        parallelogram = patches.Polygon([[0,0], [Ae1[0], Ae1[1]], 
                                         [Ae1[0]+Ae2[0], Ae1[1]+Ae2[1]], 
                                         [Ae2[0], Ae2[1]]], 
                                        closed=True, alpha=0.3, color='green', label=f'Transformed (Area={abs(matrix.determinant()):.2f})')
        ax.add_patch(parallelogram)
        
        # Plot vectors
        ax.quiver(0, 0, e1[0], e1[1], angles='xy', scale_units='xy', scale=1, color='blue', width=0.01)
        ax.quiver(0, 0, e2[0], e2[1], angles='xy', scale_units='xy', scale=1, color='blue', width=0.01)
        ax.quiver(0, 0, Ae1[0], Ae1[1], angles='xy', scale_units='xy', scale=1, color='green', width=0.015)
        ax.quiver(0, 0, Ae2[0], Ae2[1], angles='xy', scale_units='xy', scale=1, color='green', width=0.015)
        
        # Labels
        ax.text(e1[0]/2, e1[1]/2 - 0.2, 'e1', color='blue', fontsize=12, fontweight='bold')
        ax.text(e2[0]/2 - 0.2, e2[1]/2, 'e2', color='blue', fontsize=12, fontweight='bold')
        ax.text(Ae1[0], Ae1[1] + 0.2, 'Ae1', color='green', fontsize=12, fontweight='bold')
        ax.text(Ae2[0] - 0.2, Ae2[1], 'Ae2', color='green', fontsize=12, fontweight='bold')
        
        # Fixed limits so area change is obvious
        max_val = max(3, abs(Ae1[0]+Ae2[0]), abs(Ae1[1]+Ae2[1])) * 1.2
        ax.set_xlim(-max_val/2, max_val)
        ax.set_ylim(-max_val/2, max_val)
        
        ax.axhline(0, color='black', linewidth=1)
        ax.axvline(0, color='black', linewidth=1)
        ax.set_aspect('equal')
        ax.set_title(title)
        ax.legend()
        plt.show()

A1 = Matrix([[2, 0], [0, 1.5]]) # Scales area by 2 * 1.5 = 3
print("Determinant of A1:", A1.determinant())
TransformationVisualizer.plot_2d_transformation(A1, title="Area Scaled by 3 (Det=3)")

A2 = Matrix([[1, 1], [1, 1]]) # Collapses space
print("Determinant of A2:", A2.determinant())
TransformationVisualizer.plot_2d_transformation(A2, title="Dimension Collapse (Det=0)")
```

## Section 17: Matrix Inverse

If a transformation doesn't collapse space (i.e. $\det(A) \neq 0$, it has full rank), we can "undo" it using the inverse $A^{-1}$.
$$ A A^{-1} = A^{-1} A = I $$

If $\det(A) = 0$, the inverse does not exist because multiple inputs map to the same output (or collapse to the origin), meaning the transformation cannot be uniquely reversed.

## Section 18: Trace

The trace $\text{tr}(A)$ is the sum of the diagonal elements of a square matrix.

While this seems arbitrary, trace has a beautiful cyclic property: 
$$ \text{tr}(AB) = \text{tr}(BA) $$
*(assuming dimensions allow both multiplications).*

### Proof: Trace equals the sum of Eigenvalues
*(Note: We will formally prove that $A = V \Lambda V^{-1}$ in Section 23. We are using this future diagonalization result here to demonstrate the incredible power of the trace's cyclic property early on!)*

If a matrix $A$ is diagonalizable, we can write $A = V \Lambda V^{-1}$.
Using the cyclic property, let's compute the trace of $A$:

$$ \text{tr}(A) = \text{tr}(V \Lambda V^{-1}) $$

Treat $V$ as the first matrix and $(\Lambda V^{-1})$ as the second matrix. By the cyclic property, we can move $V$ to the end:

$$ \text{tr}(V (\Lambda V^{-1})) = \text{tr}((\Lambda V^{-1}) V) $$

Since $V^{-1} V = I$:
$$ \text{tr}(\Lambda I) = \text{tr}(\Lambda) $$

Because $\Lambda$ is a diagonal matrix containing the eigenvalues, its trace is simply their sum:
$$ \text{tr}(A) = \sum_{i=1}^n \lambda_i $$

Trace is invariant under a change of basis! 

### Gram Matrix and Variance
If we have a column vector $x$, the outer product is $x x^T$. By the cyclic property:
$$ \text{tr}(x x^T) = \text{tr}(x^T x) = \|x\|^2 $$
In PCA, the trace of the covariance matrix will represent the total variance in our dataset!

## Section 19: Quadratic Forms

A quadratic form maps a vector to a scalar using a square matrix $A \in \mathbb{R}^{n \times n}$:
$$ q(x) = x^T A x $$

Let's carefully derive what this actually produces for a $2 \times 2$ matrix:
$$ q(x) = \begin{bmatrix} x_1 & x_2 \end{bmatrix} \begin{bmatrix} a & b \\ c & d \end{bmatrix} \begin{bmatrix} x_1 \\ x_2 \end{bmatrix} $$

First, compute the matrix-vector multiplication $Ax$:
$$ \begin{bmatrix} a & b \\ c & d \end{bmatrix} \begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = \begin{bmatrix} a x_1 + b x_2 \\ c x_1 + d x_2 \end{bmatrix} $$

Now, multiply by the row vector $x^T$:
$$ q(x) = \begin{bmatrix} x_1 & x_2 \end{bmatrix} \begin{bmatrix} a x_1 + b x_2 \\ c x_1 + d x_2 \end{bmatrix} $$
$$ q(x) = x_1(a x_1 + b x_2) + x_2(c x_1 + d x_2) $$
$$ q(x) = a x_1^2 + b x_1 x_2 + c x_2 x_1 + d x_2^2 $$

Since $x_1 x_2 = x_2 x_1$:
$$ q(x) = a x_1^2 + (b + c) x_1 x_2 + d x_2^2 $$

**Why symmetric matrices?**
Notice that the cross-term depends entirely on $(b+c)$. For any general matrix $A$, we could replace it with a symmetric matrix $S = \frac{A + A^T}{2}$ and the quadratic form would produce the exact same scalar! 
For a symmetric matrix, $b=c$, so the cross term is simply $2b x_1 x_2$.

If $A$ is a diagonal matrix ($b=0$), then $q(x) = a x_1^2 + d x_2^2$. There are no cross terms! This perfectly represents an axis-aligned ellipse or bowl. Cross terms physically rotate that shape.

## Section 20: Matrix Definiteness Classification

Building directly on quadratic forms $q(x) = x^T A x$, we can classify symmetric matrices by the shape of their quadratic surface. 

Remember from Section 25 (which we will prove shortly) that diagonalization converts the quadratic form into a sum of squares weighted by eigenvalues:
$$ q(x) = \lambda_1 z_1^2 + \lambda_2 z_2^2 + \dots + \lambda_n z_n^2 $$

Because $z_i^2$ is always positive (or zero), the sign of the entire quadratic form depends **exclusively on the signs of the eigenvalues $\lambda_i$**.

We classify matrices as follows:
- **Positive Definite (PD)**: $q(x) > 0$ for all $x \neq 0$. 
  - All $\lambda_i > 0$. 
  - Geometrically: A strictly upward-curving bowl.
- **Positive Semidefinite (PSD)**: $q(x) \geq 0$ for all $x$. 
  - All $\lambda_i \geq 0$ (at least one is exactly 0). 
  - Geometrically: A valley (flat in at least one direction). Covariance matrices are always PSD!
- **Negative Definite (ND)**: $q(x) < 0$ for all $x \neq 0$. 
  - All $\lambda_i < 0$. 
  - Geometrically: An upside-down bowl.
- **Negative Semidefinite (NSD)**: $q(x) \leq 0$ for all $x$. 
  - All $\lambda_i \leq 0$. 
  - Geometrically: An upside-down valley.
- **Indefinite**: $q(x)$ can be positive or negative. 
  - Some $\lambda_i > 0$, some $\lambda_i < 0$. 
  - Geometrically: A saddle point (curves up in one direction, down in another).

## Section 21: Eigenvalues and Eigenvectors

Most vectors change both direction and magnitude under a transformation. **Eigenvectors** are special directions that remain on the same line! They only get scaled.

$$ A v = \lambda v $$

The scalar $\lambda$ is the eigenvalue. It determines how much the eigenvector is stretched, shrunk, or flipped.
To find them algebraically, we solve $\det(A - \lambda I) = 0$.

Let's visualize this using our library.

```python
class EigenVisualizer:
    @staticmethod
    def plot_eigenvectors(matrix, test_vectors=None, title="Eigenvectors vs Ordinary Vectors"):
        evals = matrix.eigenvalues()
        evecs = matrix.eigenvectors(evals)
        
        fig, ax = plt.subplots(figsize=(8, 8))
        
        # Plot Eigenspaces (dashed lines)
        for i, ev in enumerate(evecs):
            line_x = [ev[0] * -10, ev[0] * 10]
            line_y = [ev[1] * -10, ev[1] * 10]
            ax.plot(line_x, line_y, linestyle='--', color='gray', alpha=0.5, label=f'Eigenspace {i+1}')
            
            # Original Eigenvector v
            ax.quiver(0, 0, ev[0], ev[1], angles='xy', scale_units='xy', scale=1, color='blue', width=0.01, label=f'v{i+1}' if i==0 else None)
            ax.text(ev[0]/2, ev[1]/2 + 0.2, f'v{i+1}', color='blue', fontsize=12, fontweight='bold')
            
            # Transformed Eigenvector Av
            tev = matrix * ev
            ax.quiver(0, 0, tev[0], tev[1], angles='xy', scale_units='xy', scale=1, color='cyan', alpha=0.8, width=0.015, label=f'Av=λv' if i==0 else None)
            ax.text(tev[0], tev[1], f'Av{i+1}', color='darkcyan', fontsize=12, fontweight='bold')
            
        # Plot test vectors
        if test_vectors:
            for i, tv in enumerate(test_vectors):
                ax.quiver(0, 0, tv[0], tv[1], angles='xy', scale_units='xy', scale=1, color='red', width=0.01, label='Ordinary x' if i==0 else None)
                ttv = matrix * tv
                ax.quiver(0, 0, ttv[0], ttv[1], angles='xy', scale_units='xy', scale=1, color='orange', alpha=0.8, width=0.015, label='Ax (swings off span)' if i==0 else None)
                ax.text(tv[0], tv[1], f'x{i+1}', color='red', fontsize=12)
                ax.text(ttv[0], ttv[1], f'Ax{i+1}', color='darkorange', fontsize=12)
        
        ax.set_xlim(-4, 4)
        ax.set_ylim(-4, 4)
        ax.axhline(0, color='black')
        ax.axvline(0, color='black')
        ax.set_aspect('equal')
        ax.set_title(title)
        ax.legend(loc='upper right')
        plt.show()

# Use a symmetric matrix (guarantees orthogonal eigenvectors)
A = Matrix([[3, 1], [1, 2]])
test_vectors = [Vector([1, -1]), Vector([-1, 0.5])]

EigenVisualizer.plot_eigenvectors(A, test_vectors, title="Symmetric Matrix: Eigenspace Stretch")
```

## Section 22: Symmetric Matrices

A symmetric matrix has $A = A^T$.

For Machine Learning, symmetric matrices (like Covariance matrices) are magical because:
1. Their eigenvalues are always real.
2. Their eigenvectors are always orthogonal to each other.
3. We can always form an orthonormal basis out of their eigenvectors.

## Section 23: Diagonalization (The Conceptual Climax)

Let's derive the diagonalization formula $A = V \Lambda V^{-1}$ from scratch.

Suppose $A$ is an $n \times n$ matrix, and we find $n$ linearly independent eigenvectors $v_1, v_2, \dots, v_n$ with corresponding eigenvalues $\lambda_1, \lambda_2, \dots, \lambda_n$. 
By definition:
$$ A v_1 = \lambda_1 v_1 $$
$$ A v_2 = \lambda_2 v_2 $$
$$ \vdots $$
$$ A v_n = \lambda_n v_n $$

Let's pack all the eigenvectors into the columns of a single matrix $V = [v_1, v_2, \dots, v_n]$.
If we multiply $A$ by $V$, it applies $A$ to each column:

$$ A V = A [v_1, v_2, \dots, v_n] = [A v_1, A v_2, \dots, A v_n] $$

Substitute our eigenvalue equations:
$$ A V = [\lambda_1 v_1, \lambda_2 v_2, \dots, \lambda_n v_n] $$

Now, how can we factor the right side? We can pull the $\lambda$ scalars out into a diagonal matrix $\Lambda$ multiplying $V$ on the right:

$$ [\lambda_1 v_1, \lambda_2 v_2, \dots, \lambda_n v_n] = [v_1, v_2, \dots, v_n] \begin{bmatrix} \lambda_1 & 0 & \dots & 0 \\ 0 & \lambda_2 & \dots & 0 \\ \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & \dots & \lambda_n \end{bmatrix} = V \Lambda $$

Therefore, we have the beautiful matrix equation:
$$ A V = V \Lambda $$

If the eigenvectors are linearly independent, the matrix $V$ is invertible. Multiply both sides by $V^{-1}$ on the right:
$$ A = V \Lambda V^{-1} $$

**Geometric Interpretation:**
1. $V^{-1} x$: Change coordinates from the standard basis into the eigenvector basis.
2. $\Lambda (V^{-1} x)$: Scale each coordinate independently (no cross-terms mixing them!).
3. $V (\Lambda V^{-1} x)$: Convert the result back into the standard basis.

## Section 24: Orthogonal Diagonalization & Expansion

For a **symmetric matrix** ($A = A^T$), a mathematical miracle occurs: the eigenvectors are always orthogonal. 
If we normalize them, the eigenvector matrix $V$ becomes an orthogonal matrix $Q$. 
As we proved in Section 12, for an orthogonal matrix, $Q^{-1} = Q^T$.

Substitute this into our diagonalization formula:
$$ A = Q \Lambda Q^T $$

### Expansion into Outer Products (Spectral Decomposition)
Let's expand $Q \Lambda Q^T$ to see what $A$ actually *is*. 
$$ Q = [q_1, q_2, \dots, q_n] $$
$$ \Lambda = \text{diag}(\lambda_1, \lambda_2, \dots, \lambda_n) $$
$$ Q^T = \begin{bmatrix} q_1^T \\ q_2^T \\ \vdots \\ q_n^T \end{bmatrix} $$

If you carry out the block matrix multiplication, you get a sum of rank-1 outer products:
$$ A = \sum_{i=1}^n \lambda_i q_i q_i^T $$

**What does this mean geometrically?**
If we apply $A$ to a vector $x$:
$$ Ax = \sum_{i=1}^n \lambda_i (q_i q_i^T) x = \sum_{i=1}^n \lambda_i q_i (q_i^T x) $$
Notice that $q_i^T x$ is just a scalar (the dot product/projection of $x$ onto $q_i$). 
So $A$ breaks $x$ down into its projections along each eigenvector $q_i$, scales each projection by $\lambda_i$, and adds them all back together!

## Section 25: Unifying Quadratic Forms and Eigenvalues

Let's revisit our quadratic form $q(x) = x^T A x$.

If $A$ is symmetric, we can orthogonally diagonalize it: $A = Q \Lambda Q^T$.
Let's substitute this directly into the quadratic form:
$$ q(x) = x^T (Q \Lambda Q^T) x $$

Because matrix multiplication is associative, we can group the terms:
$$ q(x) = (x^T Q) \Lambda (Q^T x) $$

Notice that $x^T Q = (Q^T x)^T$. Let's define a new vector $z = Q^T x$. 
(Geometrically, $z$ represents the coordinates of $x$ in our new eigenvector basis).

Substitute $z$:
$$ q(x) = z^T \Lambda z $$

Since $\Lambda$ is a diagonal matrix, this expands beautifully:
$$ q(x) = \begin{bmatrix} z_1 & z_2 & \dots & z_n \end{bmatrix} \begin{bmatrix} \lambda_1 & 0 & \dots & 0 \\ 0 & \lambda_2 & \dots & 0 \\ \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & \dots & \lambda_n \end{bmatrix} \begin{bmatrix} z_1 \\ z_2 \\ \vdots \\ z_n \end{bmatrix} $$

$$ q(x) = \lambda_1 z_1^2 + \lambda_2 z_2^2 + \dots + \lambda_n z_n^2 $$

**The cross terms ($x_1 x_2$) are completely gone!** 
Diagonalization rotated our perspective so that the quadratic form's principal axes line up perfectly with our coordinate axes. This is the mathematical mechanism that makes PCA work: finding the rotation $Q$ that decouples our correlated variables.

## Section 26: Preparation for PCA

Everything in this notebook is the theoretical foundation for Principal Component Analysis (PCA).

The steps of PCA will mirror exactly what we have built:
1. **Data Matrix $X$**: Collect your data.
2. **Center the data**: Subtract the mean.
3. **Covariance Matrix $C$**: Compute $C$. It is symmetric and positive semidefinite!
4. **Orthogonal Diagonalization**: Because $C$ is symmetric, $C = Q \Lambda Q^T$.
5. **Principal Directions**: The columns of $Q$ are orthonormal eigenvectors. They represent the axes of greatest variance.
6. **Variance**: The eigenvalues in $\Lambda$ represent the amount of variance along each direction. The trace $\sum \lambda_i$ is the total variance.
7. **Change of Basis (Projection)**: $z = Q^T x$. 
8. **Dimensionality Reduction**: Keep only the largest coordinates in $z$.
9. **Reconstruction**: $\hat{x} = Q_k z + \text{mean}$.

We have successfully built the mathematical machinery to understand PCA from the ground up.
